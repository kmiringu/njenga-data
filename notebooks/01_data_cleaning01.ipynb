```python
# cleaning cahf_benchmarks
## importing libraries

import pandas as pd 
import numpy as np
```

```python
## loading
df_cahf = pd.read_csv("../data/raw/Cahf_benchmarks.csv", encoding= "utf-8")
df_cahf.info()
df_cahf.describe()
df_cahf.shape

df_cahf
```

```python
# cleaning data types
df_cahf["amount_kes"] = df_cahf["amount_kes"].astype(float)

# Clean null values
df_cahf.isnull().sum() # no null values present

# Drop columns
df_cahf = df_cahf.drop(columns= ["source_bucket"])

# Stripping white spaces
df_cahf["county"] = df_cahf["county"].str.strip()
df_cahf["unit_type"] = df_cahf["unit_type"].str.strip()
df_cahf["cost_category"] = df_cahf["cost_category"].str.strip()
```
